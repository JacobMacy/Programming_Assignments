```python
#13.1
from datetime import date, datetime

today = date.today()
with open("today.txt", "w", encoding="utf-8") as file:
    file.write(today.isoformat())

print("Date written:", today.isoformat())
#13.2
with open("today.txt", "r", encoding="utf-8") as file:
    today_string = file.read()

print("today_string:", repr(today_string))
print("Type:", type(today_string).__name__)
#13.3
parsed_date = datetime.strptime(today_string.strip(), "%Y-%m-%d").date()

print("Parsed date:", parsed_date)
print("Type:", type(parsed_date).__name__)
print("Year:", parsed_date.year)
print("Month:", parsed_date.month)
print("Day:", parsed_date.day)
#15.1
from pathlib import Path
import subprocess
import sys

process_script = r'''import multiprocessing
import random
import time
from datetime import datetime


def worker(number):
    delay = random.uniform(0, 1)
    time.sleep(delay)
    print(
        f"Process {number}: waited {delay:.3f} seconds; "
        f"current time: {datetime.now().strftime('%H:%M:%S.%f')}",
        flush=True,
    )


if __name__ == "__main__":
    processes = [
        multiprocessing.Process(target=worker, args=(number,))
        for number in range(1, 4)
    ]
    for process in processes:
        process.start()
    for process in processes:
        process.join()
    if any(process.exitcode != 0 for process in processes):
        raise RuntimeError("A worker process failed.")
'''

script_path = Path("exercise_15_1.py")
script_path.write_text(process_script, encoding="utf-8")
result = subprocess.run(
    [sys.executable, str(script_path.resolve())],
    capture_output=True,
    text=True,
    check=True,
)
print(result.stdout, end="")
```
